```sql
CREATE OR REPLACE TABLE 
workspace.default.transformed_sales
USING DELTA
AS
SELECT
    -- Order information
    oc.order_id,
    oi.order_item_id,
    oc.order_status,

    -- Customer
    oc.customer_id,
    oc.customer_unique_id,
    oc.customer_city,
    oc.customer_state,
    oc.customer_zip_code_prefix,

    -- Product
    oi.product_id,
    p.product_category_name,
    p.product_category_name_english,
    p.product_name_lenght,
    p.product_description_lenght,
    p.product_photos_qty,
    p.product_weight_g,
    p.product_length_cm,
    p.product_height_cm,
    p.product_width_cm,

    -- Seller
    oi.seller_id,
    s.seller_city,
    s.seller_state,
    s.seller_zip_code_prefix,

    -- Dates
    oc.order_date,
    oc.order_purchase_timestamp,
    oc.order_year,
    oc.order_month,
    oc.order_month_name,
    oc.order_quarter,
    oc.order_day,
    oc.order_day_name,
    oc.order_estimated_delivery_date,

    -- Delivery
    oc.delivery_days,
    oc.delivery_performance,

    -- Item financials
    oi.price,
    oi.freight_value,
    oi.item_sales,
    oi.shipping_cost,
    oi.total_item_value,

    -- Order financials
    r.product_revenue,
    r.shipping_revenue,
    r.order_total_value,
    r.item_count,
    r.unique_products,
    r.unique_sellers,

    -- Payments
    pay.total_payment,
    pay.payment_count,
    pay.max_installments,
    pp.primary_payment_type,
    pp.primary_payment_installments,

    -- Payment difference
    ROUND(
        COALESCE(pay.total_payment, 0)
        -
        COALESCE(r.order_total_value, 0),
        2
    ) AS payment_difference

FROM workspace.default.transformed_order_customer oc

LEFT JOIN
    workspace.default.transformed_order_items oi
ON
    oc.order_id = oi.order_id
LEFT JOIN
    workspace.default.transformed_products p
ON
    oi.product_id = p.product_id
LEFT JOIN
    workspace.default.transformed_order_revenue r
ON
    oc.order_id = r.order_id
LEFT JOIN
    workspace.default.transformed_payments pay
ON
    oc.order_id = pay.order_id
LEFT JOIN
    workspace.default.transformed_primary_payment pp
ON
    oc.order_id = pp.order_id
LEFT JOIN
    workspace.default.silver_sellers s
ON
    oi.seller_id = s.seller_id;
```

```sql
create or replace table 
workspace.default.transformed_order_customer
using delta as
select
o.order_id,
o.customer_id,
c.customer_unique_id,
c.customer_zip_code_prefix,
c.customer_city,
c.customer_state,
o.order_status,
o.order_purchase_timestamp,
o.order_approved_at,
o.order_delivered_carrier_date,
o.order_delivered_customer_date,
o.order_estimated_delivery_date,
o.order_date,
o.order_year,
o.order_month,
o.order_month_name,
o.order_quarter,
o.order_day,
o.order_day_name,
o.delivery_days,
o.delivery_performance
from workspace.default.transformed_orders o 
left join workspace.default.silver_customers c
on o.customer_id=c.customer_id;
```

```sql
create or replace table
workspace.default.transformed_primary_payment
using delta as
with ranked_payments as(
    select order_id,
    payment_type,
    payment_value,
    payment_installments,
    row_number() over(
        partition by order_id
        order by payment_value desc
    ) as payment_rank
    from workspace.default.silver_payments
)

select order_id,
payment_type as primary_payment_type,
payment_installments as primary_payment_installments
from ranked_payments
where payment_rank = 1
```

```sql
create or replace table
workspace.default.transformed_payments
using delta as
select
order_id,
ROUND(
    SUM(payment_value),
    2
) AS total_payment,
COUNT(payment_sequential)
    AS payment_count,
    MAX(payment_installments)
    AS max_installments
FROM workspace.default.silver_payments
GROUP BY order_id;
```

```sql
create or replace table
workspace.default.transformed_order_revenue
using delta as
select
order_id,
round(sum(item_sales),2) as product_revenue,
round(sum(shipping_cost),2) as shipping_revenue,
round(sum(total_item_value),2) as order_total_value,
count(order_item_id) as item_count,
count(distinct product_id) as unique_products,
count(distinct seller_id) as unique_sellers
from workspace.default.transformed_order_items
group by order_id;
```

```sql
create or replace table
workspace.default.transformed_order_items
using delta as
select 
order_id,
order_item_id,
product_id,
seller_id,
shipping_limit_date,
price,
freight_value,

round(price,2) as item_sales,
round(freight_value,2) as shipping_cost,
round(price+freight_value,2) as total_item_value,
case 
when price < 50
then 'low'
when price < 200
then 'medium'
else 'high'
end as revenue_category
FROM workspace.default.silver_order_items;
```

```sql
create or replace table
workspace.default.transformed_orders
using delta as
select order_id,
customer_id,
order_status,
order_purchase_timestamp,
order_approved_at,
order_delivered_carrier_date,
order_delivered_customer_date,
order_estimated_delivery_date,
cast(order_purchase_timestamp as DATE) as order_date,
YEAR(order_purchase_timestamp) AS order_year,
MONTH(order_purchase_timestamp) AS order_month,
QUARTER(order_purchase_timestamp) AS order_quarter,
DAY(order_purchase_timestamp) AS order_day,
DAYOFWEEK(order_purchase_timestamp) AS order_dayofweek,
DATE_FORMAT(
    order_purchase_timestamp,
    'MMMM'
) AS order_month_name,
DATE_FORMAT(
    order_purchase_timestamp,
    'EEEE'
) AS order_day_name,
CASE
    WHEN order_delivered_customer_date IS NOT NULL
    THEN DATEDIFF(
        order_delivered_customer_date,
        order_purchase_timestamp
    )
    ELSE NULL
END AS delivery_days,
CASE

    WHEN order_delivered_customer_date IS NULL
        THEN 'not_delivered'

    WHEN order_delivered_customer_date >
         order_estimated_delivery_date
        THEN 'late'

    ELSE 'on_time'

END AS delivery_performance
FROM workspace.default.silver_orders;
```
