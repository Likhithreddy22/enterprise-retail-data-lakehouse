```python
from pyspark.sql.functions import *
```

```python
DATABASE = "workspace.default"
```

```python
customers = spark.table("workspace.default.silver_customers")

orders = spark.table("workspace.default.silver_orders")

order_items = spark.table("workspace.default.silver_order_items")

products = spark.table("workspace.default.silver_products")

sellers = spark.table("workspace.default.silver_sellers")

payments = spark.table("workspace.default.silver_payments")

reviews = spark.table("workspace.default.silver_reviews")

geolocation = spark.table("workspace.default.silver_geolocation")

translation = spark.table("workspace.default.silver_category_translation")
```

```python
duplicate_customers = (
    customers.groupBy("customer_id")
    .count()
    .filter(col("count") > 1)
)

display(duplicate_customers)
```

```python
duplicate_orders = (
    orders.groupBy("order_id")
    .count()
    .filter(col("count") > 1)
)

display(duplicate_orders)
```

```python
missing_customer = customers.filter(
    col("customer_id").isNull()
)

display(missing_customer)
```

```python
invalid_price = order_items.filter(
    col("price") <= 0
)

display(invalid_price)
```

```python
invalid_freight = order_items.filter(
    col("freight_value") < 0
)

display(invalid_freight)
```

```python
invalid_weight = products.filter(
    col("product_weight_g") < 0
)

display(invalid_weight)
```

```python
invalid_review_score = reviews.filter(
    ~col("review_score").between(1,5)
)

display(invalid_review_score)
```

```python
invalid_payment = payments.filter(
    col("payment_value") <= 0
)

display(invalid_payment)
```

```python
invalid_installments = payments.filter(
    col("payment_installments") < 0
)

display(invalid_installments)
```

```python
invalid_customer_orders = (
    orders.join(
        customers,
        "customer_id",
        "left_anti"
    )
)

display(invalid_customer_orders)
```

```python
invalid_products = (
    order_items.join(
        products,
        "product_id",
        "left_anti"
    )
)

display(invalid_products)
```

```python
invalid_sellers = (
    order_items.join(
        sellers,
        "seller_id",
        "left_anti"
    )
)

display(invalid_sellers)
```

```python
invalid_delivery = orders.filter(
    (col("order_status") == "delivered") &
    (col("order_delivered_customer_date").isNull())
)

display(invalid_delivery)
```

```python
invalid_dates = orders.filter(
    col("order_delivered_customer_date") <
    col("order_purchase_timestamp")
)

display(invalid_dates)
```

```python
invalid_delivery.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable("workspace.default.rejected_delivery")
```

```python
summary = [
    ("Duplicate Customers", duplicate_customers.count()),
    ("Duplicate Orders", duplicate_orders.count()),
    ("Missing Customers", missing_customer.count()),
    ("Invalid Prices", invalid_price.count()),
    ("Invalid Freight", invalid_freight.count()),
    ("Invalid Product Weight", invalid_weight.count()),
    ("Invalid Review Score", invalid_review_score.count()),
    ("Invalid Payment", invalid_payment.count()),
    ("Invalid Installments", invalid_installments.count()),
    ("Invalid Customer Orders", invalid_customer_orders.count()),
    ("Invalid Products", invalid_products.count()),
    ("Invalid Sellers", invalid_sellers.count()),
    ("Invalid Delivery", invalid_delivery.count()),
    ("Invalid Dates", invalid_dates.count())
]

validation_report = spark.createDataFrame(
    summary,
    ["validation", "failed_records"]
)

display(validation_report)
```

```python
validation_report.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable("workspace.default.validation_report")
```
