```python
def load_to_postgres(df, table_name):

    print(f"Loading {table_name}...")

    (
        df.write
        .format("postgresql")
        .option("host", pg_host)
        .option("port", pg_port)
        .option("database", pg_database)
        .option("user", pg_user)
        .option("password", pg_password)
        .option("dbtable", f"retail.{table_name}")
        .mode("overwrite")
        .save()
    )

    print(f"{table_name} loaded successfully.")
```

```python
dim_payment = spark.table(
    "workspace.default.dim_payment"
)

print("Rows:", dim_payment.count())

load_to_postgres(
    dim_payment,
    "dim_payment"
)
```
